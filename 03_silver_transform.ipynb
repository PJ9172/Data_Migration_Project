```python
from pyspark.sql.functions import col
```

```python
orders_raw = spark.table("demo_catalog.bronze.orders_raw")
```

```python
valid = orders_raw.filter(
    col("order_id").isNotNull() & col("customer_id").isNotNull() &
    col("order_status").isin("PLACED","SHIPPED","DELIVERED","CANCELLED")
)
rejected = orders_raw.exceptAll(valid)
```

```python
valid.write.mode("overwrite").saveAsTable("demo_catalog.silver.orders")
rejected.write.mode("overwrite").saveAsTable("demo_catalog.silver.orders_rejected")
```

```python
for t in ["customers", "products", "order_items"]:
    spark.table(f"demo_catalog.bronze.{t}_raw").write.mode("overwrite").saveAsTable(f"demo_catalog.silver.{t}")
```
